```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier  
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import classification_report, roc_auc_score
```

df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
df.info()
df.head()

```python
label_encoder = LabelEncoder()
for colum in df.select_dtypes(include=['object']).columns:
    if colum != 'Churn':
        df[colum] = label_encoder.fit_transform(df[colum])
```

```python
df['Churn'] = label_encoder.fit_transform(df['Churn'])
df.head()
```

```python
df.info()
```

```python
scaler = StandardScaler()
numerical_features = ['tenure', 'MonthlyCharges', 'TotalCharges']
df[numerical_features]= scaler.fit_transform(df[numerical_features])
df
```

```python
x = df.drop(columns=['Churn'])
y = df['Churn']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(pd.Series(y_train).value_counts())
```

```python
smote = SMOTE(random_state=42)
x_train_smote, y_train_smote = smote.fit_resample(x_train, y_train)
print(pd.Series(y_train_smote).value_counts())
```

```python
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(x_train, y_train)
pred_rf_model = rf_model.predict(x_test)
print(classification_report(y_test, pred_rf_model))
print(roc_auc_score(y_test, pred_rf_model))
```

```python
rf_model_clas = RandomForestClassifier(random_state=42)
rf_model_clas.fit(x_train_smote, y_train_smote)
pred_rf_model_clas = rf_model_clas.predict(x_test)
print(classification_report(y_test, pred_rf_model_clas))
print(roc_auc_score(y_test, rf_model_clas.predict_proba(x_test)[:, 1]))
```

```python
xg_model = XGBClassifier(eval_metric='logloss', random_state=42)
xg_model.fit(x_train, y_train)
pred_xg = xg_model.predict(x_test)
print(classification_report(y_test, pred_xg))
print(roc_auc_score(y_test, xg_model.predict_proba(x_test)[:, 1]))
```

```python
xg_model_smote = XGBClassifier(eval_metric='logloss', random_state=42)
xg_model_smote.fit(x_train_smote, y_train_smote)
pred_xg_smote = xg_model_smote.predict(x_test)
print(classification_report(y_test, pred_xg_smote))
print(roc_auc_score(y_test, xg_model_smote.predict_proba(x_test)[:, 1]))
```

```python
lgb_model = LGBMClassifier(random_state=42)
lgb_model.fit(x_train, y_train)
pred_lgb = lgb_model.predict(x_test)
print(classification_report(y_test, pred_lgb))
print(roc_auc_score(y_test, lgb_model.predict_proba(x_test)[:, 1]))
```

```python
lgb_model_smote = LGBMClassifier(random_state=42)
lgb_model_smote.fit(x_train_smote, y_train_smote)
pred_lgb_model_smote = lgb_model_smote.predict(x_test)
print(classification_report(y_test, pred_lgb_model_smote))
print(roc_auc_score(y_test, lgb_model_smote.predict_proba(x_test)[:, 1]))
```
